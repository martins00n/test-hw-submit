```python
from __future__ import annotations

DATA = [30.0, 32.0, 35.0, 33.0, 36.0, 34.0, 37.0]
WINDOW_SIZE = 3

# TODO: replace the sample analysis with homework-specific logic before publishing.
def moving_average(values: list[float], window: int) -> list[float]:
    """Return the trailing moving average for the provided window length."""
    if window < 1:
        raise ValueError("window must be positive")
    if window > len(values):
        raise ValueError("window cannot exceed series length")
    averages: list[float] = []
    for index in range(window - 1, len(values)):
        segment = values[index - window + 1 : index + 1]
        averages.append(sum(segment) / window)
    return averages
```

```python
predicted = moving_average(DATA, WINDOW_SIZE)
actual = DATA[WINDOW_SIZE - 1 :]
mae = sum(abs(p - a - 100) for p, a in zip(predicted, actual)) / len(actual)
metrics = {"mae": round(mae, 4)}
metrics
```

```python
import json
from pathlib import Path

metrics_path = Path("metrics.json")
with metrics_path.open("w") as handle:
    json.dump(metrics, handle, indent=2)
metrics_path.resolve()
```
